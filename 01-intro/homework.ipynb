```python
# import library
import numpy as np
import pandas as pd
```

```python
# import dataset
df = pd.read_csv("car_fuel_efficiency_2026.csv")
```

```python
# Q1. check pandas version
print('My pandas version:', pd.__version__)
```

```python
# Q2. records count
print('Number of records in the dataset:', len(df))
```

```python
# Q3. fuel types
fuel_type = df['fuel_type'].unique()
count_fuel_type = len(fuel_type)
print('The type of fuel present in the dataset:', fuel_type.tolist())
print('The number of fuel types presented in the dataset:', count_fuel_type)
```

```python
# Q4. missing values
missing_value = (df.isnull().sum() > 0).sum()
print('The number of columns in the dataset that have missing values:', missing_value)
```

```python
# Q5. max fuel efficiency
car_asia = df[df['origin'] == 'Asia']
max_fuel_efficiency = car_asia['fuel_efficiency_mpg'].max()
print('Maximum fuel efficiency for cars from asia:', max_fuel_efficiency)
```

```python
# Q6. median value of horsepower
median_horsepower = df['horsepower'].median()
frequent_value_horsepower = df['horsepower'].mode()[0]
fill_miss_value = df['horsepower'].fillna(frequent_value_horsepower)
median_horsepower_now = df['horsepower'].median()
print('Median of the horsepower column:', median_horsepower)
print('Frequent value of the horsepower column:', frequent_value_horsepower)
print('Median of the horsepower column after fill missing value with Frequent value:', median_horsepower_now)

print('Has median changed?:')
if median_horsepower_now != median_horsepower:
    if median_horsepower_now > median_horsepower:
        print('Yes, it increased')
    else:
        print('Yes, it decreased')
else:
    print('No')
```

```python
# Q7. sum of weights
X = car_asia[['vehicle_weight', 'model_year']].head(7).to_numpy()
XTX = X.T @ X
XTX_invers = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_invers @ X.T @ y
w_sum = np.sum(w)
print('The sum of all elements:', w_sum)
print('Answer: ', round(w_sum, 3))
```
